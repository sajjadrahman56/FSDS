### Different types of Exception Handling 

```python
try:
    with open("files/sajjad.txt","r") as f:
        data = f.read()
        print(data)
except FileNotFoundError:
    print("File name is not correct , if not present create it first")
except NameError:
    print("Variable is not defined")
except IndexError:
    print("List index is out of range")
    
```

```python
# we can simply use it like this 

try:
    with open("files/sajjad.txt","r") as f:
        data = f.read()
        print(data)
except Exception as e:
    print("An error occurred:", e)
```
