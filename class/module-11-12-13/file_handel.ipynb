### File Handeling context Manager (with)

Manually we have to close the file after read or write but here we have to use an updated version `with` 
- If we do not close it, garbage collector would close it 
- with keyword closes the file as soon as done

```python
with open('files/demo.txt','w') as f:
    f.write("yes it is new text")
```

```python
f.write("is the file is open")
```

```python
with open('logo.png','r') as f:
    print(f.read())
```

```python
with open('logo.png','rb') as f:
    print(f.read())
```

- Text file has some limitations such as we cannot store other data types here instead of the STRING data type. 

- JSON is the universal data model types . 

- Serialization : Python to JSON convert
- Deserialization : JSON to Python Convert

```python
import json 

L = [1,2,3,4,5,6]

with open("files/test.json","w") as f:
    json.dump(L,f)
```

```python
with open("files/test.json","r") as f:
    data = json.load(f)
    print(data)
    print(type(data))
```

### Pickling

Pickling is the process whereby a Python is converted into a byte stream , and unpickling is the operation whereby a byte stream is converted back into a object hierarchy


```python
def I_am():
    return "I am Sajjad"

d = I_am()
```

```python
type(d)
```

```python
import pickle

with open("files/funtions.pkl","wb") as p:
    pickle.dump(d,p)
```

```python
with open("files/funtions.pkl","rb") as p:
    f = pickle.load(p)
    print(f)
```
